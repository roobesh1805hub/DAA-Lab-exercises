```python
import math

# ---------- First Fit ----------
def first_fit(items, capacity):
    bins = []
    contents = []

    for item in items:
        placed = False

        for i in range(len(bins)):
            if bins[i] >= item:
                bins[i] -= item
                contents[i].append(item)
                placed = True
                break

        if not placed:
            bins.append(capacity - item)
            contents.append([item])

    return contents


# ---------- First Fit Decreasing ----------
def first_fit_decreasing(items, capacity):
    return first_fit(sorted(items, reverse=True), capacity)


# ---------- Best Fit Decreasing ----------
def best_fit_decreasing(items, capacity):
    items = sorted(items, reverse=True)

    bins = []
    contents = []

    for item in items:

        best = -1
        min_space = float("inf")

        for i in range(len(bins)):
            if bins[i] >= item and bins[i] - item < min_space:
                best = i
                min_space = bins[i] - item

        if best == -1:
            bins.append(capacity - item)
            contents.append([item])
        else:
            bins[best] -= item
            contents[best].append(item)

    return contents


def display(name, bins):
    print(f"\n{name}")
    print("Servers Used:", len(bins))

    for i, server in enumerate(bins, start=1):
        print(f"Server {i}: {server}  Used = {sum(server):.1f} GB")


# ---------------- Driver ----------------

capacity = 8

tasks = [
    1.5, 2.0, 3.5, 0.5, 4.0,
    1.0, 2.5, 3.0, 1.5, 2.0,
    0.5, 4.0, 1.0, 3.5, 2.0
]

print("Total RAM Required =", sum(tasks), "GB")
print("Lower Bound =", math.ceil(sum(tasks) / capacity), "servers")

ffd = first_fit_decreasing(tasks, capacity)
bfd = best_fit_decreasing(tasks, capacity)

display("First Fit Decreasing (FFD)", ffd)
display("Best Fit Decreasing (BFD)", bfd)

print("\nSummary")
print("FFD Servers =", len(ffd))
print("BFD Servers =", len(bfd))
```
