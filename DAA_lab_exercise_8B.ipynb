```python
from itertools import permutations

INF = float('inf')

# ---------- Brute Force TSP ----------
def tsp_brute_force(cost):
    n = len(cost)
    cities = range(1, n)

    best_cost = INF
    best_path = None

    for perm in permutations(cities):
        path = [0] + list(perm) + [0]

        total = 0
        for i in range(n):
            total += cost[path[i]][path[i + 1]]

        if total < best_cost:
            best_cost = total
            best_path = path

    return best_path, best_cost


# ---------- Nearest Neighbour ----------
def nearest_neighbour(cost, start=0):
    n = len(cost)

    visited = [False] * n
    visited[start] = True

    path = [start]
    total = 0
    current = start

    for _ in range(n - 1):

        next_city = -1
        min_cost = INF

        for city in range(n):
            if not visited[city] and cost[current][city] < min_cost:
                min_cost = cost[current][city]
                next_city = city

        visited[next_city] = True
        path.append(next_city)
        total += min_cost
        current = next_city

    total += cost[current][start]
    path.append(start)

    return path, total


# ---------- Cost Matrix (Road Distances in km) ----------
cities = ["A", "B", "C", "D", "E", "F"]

cost = [
    [INF, 12, 10, 19, 8, 14],
    [12, INF, 3, 7, 2, 10],
    [10, 3, INF, 6, 20, 4],
    [19, 7, 6, INF, 4, 5],
    [8, 2, 20, 4, INF, 9],
    [14, 10, 4, 5, 9, INF]
]

# ---------- Solve ----------
opt_path, opt_cost = tsp_brute_force(cost)
nn_path, nn_cost = nearest_neighbour(cost)

print("Optimal Tour (Brute Force)")
print(" -> ".join(cities[i] for i in opt_path))
print("Cost =", opt_cost)

print("\nNearest Neighbour Tour")
print(" -> ".join(cities[i] for i in nn_path))
print("Cost =", nn_cost)
```
