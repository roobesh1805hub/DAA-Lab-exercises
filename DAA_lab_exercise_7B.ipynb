```python
def solve_n_queens_bitmask(n):
    solutions = []
    backtracks = 0

    def backtrack(row, cols, diag1, diag2, board):
        nonlocal backtracks

        if row == n:
            solutions.append(board[:])
            return

        # Bits representing free columns
        available = ((1 << n) - 1) & ~(cols | diag1 | diag2)

        while available:
            # Rightmost available position
            bit = available & -available
            available -= bit

            col = bit.bit_length() - 1
            board[row] = col

            backtrack(
                row + 1,
                cols | bit,
                (diag1 | bit) << 1,
                (diag2 | bit) >> 1,
                board
            )

            board[row] = -1
            backtracks += 1

    board = [-1] * n
    backtrack(0, 0, 0, 0, board)

    return len(solutions), backtracks


# Driver Code
n = 12

solutions, backtracks = solve_n_queens_bitmask(n)

print("N =", n)
print("Total Solutions :", solutions)
print("Total Backtracks:", backtracks)
```
