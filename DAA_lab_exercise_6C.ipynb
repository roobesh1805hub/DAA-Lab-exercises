```python
import time

# ---------------- Bottom-Up DP ----------------

def matrix_chain_bottom_up(dims):
    n = len(dims) - 1

    m = [[0] * (n + 1) for _ in range(n + 1)]

    for l in range(2, n + 1):
        for i in range(1, n - l + 2):
            j = i + l - 1
            m[i][j] = float('inf')

            for k in range(i, j):
                cost = (
                    m[i][k]
                    + m[k + 1][j]
                    + dims[i - 1] * dims[k] * dims[j]
                )

                if cost < m[i][j]:
                    m[i][j] = cost

    return m[1][n]


# ---------------- Top-Down Memoization ----------------

def matrix_chain_memo(dims):
    n = len(dims) - 1

    memo = [[None] * (n + 1) for _ in range(n + 1)]

    def solve(i, j):
        if i == j:
            return 0

        if memo[i][j] is not None:
            return memo[i][j]

        memo[i][j] = float('inf')

        for k in range(i, j):
            cost = (
                solve(i, k)
                + solve(k + 1, j)
                + dims[i - 1] * dims[k] * dims[j]
            )

            memo[i][j] = min(memo[i][j], cost)

        return memo[i][j]

    return solve(1, n)


# Driver Code (6 Matrices)

dims = [30, 35, 15, 5, 10, 20, 25]

print("Dimensions:", dims)

# Bottom-Up
start = time.time()
bottom_cost = matrix_chain_bottom_up(dims)
bottom_time = time.time() - start

# Top-Down
start = time.time()
memo_cost = matrix_chain_memo(dims)
memo_time = time.time() - start

print("\nBottom-Up DP Cost :", bottom_cost)
print("Top-Down Memo Cost:", memo_cost)

print("\nBottom-Up Time : {:.6f} seconds".format(bottom_time))
print("Top-Down Time  : {:.6f} seconds".format(memo_time))

if bottom_cost == memo_cost:
    print("\nVerification Successful: Both methods give the same minimum cost.")
else:
    print("\nVerification Failed!")
```
