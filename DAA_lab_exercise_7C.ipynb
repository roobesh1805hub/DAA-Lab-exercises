```python
N = 8

moves_x = [2, 1, -1, -2, -2, -1, 1, 2]
moves_y = [1, 2, 2, 1, -1, -2, -2, -1]


def is_safe(x, y, board):
    return (
        0 <= x < N and
        0 <= y < N and
        board[x][y] == -1
    )


def solve_knight(board, x, y, move_no):

    if move_no == N * N:
        return True

    for i in range(8):
        nx = x + moves_x[i]
        ny = y + moves_y[i]

        if is_safe(nx, ny, board):
            board[nx][ny] = move_no

            if solve_knight(board, nx, ny, move_no + 1):
                return True

            # Backtrack
            board[nx][ny] = -1

    return False


def knight_tour():
    board = [[-1 for _ in range(N)] for _ in range(N)]

    board[0][0] = 0

    if solve_knight(board, 0, 0, 1):
        print("Knight's Tour Solution:\n")

        for row in board:
            for cell in row:
                print(f"{cell:2}", end=" ")
            print()
    else:
        print("No solution exists.")


# Driver Code
knight_tour()
```
