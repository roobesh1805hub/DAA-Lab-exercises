```python
import random
import math
from itertools import permutations

INF = float("inf")


# ---------- Distance ----------
def distance(a, b):
    return round(math.sqrt((a[0]-b[0])**2 + (a[1]-b[1])**2), 2)


# ---------- Build Distance Matrix ----------
def build_matrix(points):
    n = len(points)

    cost = [[0]*n for _ in range(n)]

    for i in range(n):
        for j in range(n):
            if i == j:
                cost[i][j] = INF
            else:
                cost[i][j] = distance(points[i], points[j])

    return cost


# ---------- Tour Cost ----------
def tour_cost(path, cost):
    total = 0

    for i in range(len(path)-1):
        total += cost[path[i]][path[i+1]]

    return total


# ---------- Nearest Neighbour ----------
def nearest_neighbour(cost):

    n = len(cost)

    visited = [False]*n
    visited[0] = True

    path = [0]
    current = 0

    while len(path) < n:

        best = None
        best_dist = INF

        for city in range(n):
            if not visited[city] and cost[current][city] < best_dist:
                best = city
                best_dist = cost[current][city]

        visited[best] = True
        path.append(best)
        current = best

    path.append(0)

    return path


# ---------- 2-opt ----------
def two_opt(path, cost):

    improved = True

    while improved:
        improved = False

        for i in range(1, len(path)-2):

            for j in range(i+1, len(path)-1):

                if j-i == 1:
                    continue

                new_path = path[:]
                new_path[i:j] = reversed(path[i:j])

                if tour_cost(new_path, cost) < tour_cost(path, cost):
                    path = new_path
                    improved = True

    return path


# ---------- Brute Force (10 cities is slow) ----------
def brute_force(cost):

    n = len(cost)

    best_cost = INF
    best_path = None

    for perm in permutations(range(1, n)):
        path = [0] + list(perm) + [0]

        c = tour_cost(path, cost)

        if c < best_cost:
            best_cost = c
            best_path = path

    return best_path, best_cost


# ---------- Driver ----------

random.seed(10)

points = [(random.randint(0,100), random.randint(0,100)) for _ in range(10)]

cost = build_matrix(points)

nn_path = nearest_neighbour(cost)
nn_cost = tour_cost(nn_path, cost)

opt_path = two_opt(nn_path, cost)
opt_cost = tour_cost(opt_path, cost)

print("Nearest Neighbour Cost :", round(nn_cost,2))
print("2-opt Improved Cost    :", round(opt_cost,2))

# NOTE:
# Brute-force for 10 cities requires evaluating 9! = 362,880 tours,
# which is computationally expensive.
# Uncomment below if you want the exact optimal solution.

# bf_path, bf_cost = brute_force(cost)
# print("Optimal Cost (Brute Force):", round(bf_cost,2))
# print("2-opt Error:", round(((opt_cost-bf_cost)/bf_cost)*100,2), "%")
```
