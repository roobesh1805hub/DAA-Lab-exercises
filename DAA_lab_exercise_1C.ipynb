```python
import random

def interpolation_search_float(arr, target):
    low, high = 0, len(arr) - 1
    comparisons = 0

    while low <= high and arr[low] <= target <= arr[high]:
        comparisons += 1

        pos = low + int(
            ((target - arr[low]) * (high - low))
            / (arr[high] - arr[low])
        )

        if abs(arr[pos] - target) < 0.0001:
            return pos, comparisons
        elif arr[pos] < target:
            low = pos + 1
        else:
            high = pos - 1

    return -1, comparisons


sizes = [10000, 50000, 100000]

print("Size\tComparisons")

for size in sizes:
    arr = sorted(
        [random.uniform(0.0, 1000.0) for _ in range(size)]
    )

    target = arr[size // 2]

    idx, comp = interpolation_search_float(arr, target)

    print(size, "\t", comp)
```
