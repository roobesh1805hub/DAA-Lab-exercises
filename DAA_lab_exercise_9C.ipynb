```python
import random

CAPACITY = 1.0


# ---------- Next Fit ----------
def next_fit(items, capacity=1.0):
    bins = 1
    remaining = capacity

    for item in items:
        if item <= remaining:
            remaining -= item
        else:
            bins += 1
            remaining = capacity - item

    return bins


# ---------- Next Fit Decreasing ----------
def next_fit_decreasing(items, capacity=1.0):
    items = sorted(items, reverse=True)
    return next_fit(items, capacity)


# ---------- First Fit ----------
def first_fit(items, capacity=1.0):
    space = []

    for item in items:
        placed = False

        for i in range(len(space)):
            if space[i] >= item:
                space[i] -= item
                placed = True
                break

        if not placed:
            space.append(capacity - item)

    return len(space)


# ---------- First Fit Decreasing ----------
def first_fit_decreasing(items, capacity=1.0):
    return first_fit(sorted(items, reverse=True), capacity)


# ---------- Best Fit Decreasing ----------
def best_fit_decreasing(items, capacity=1.0):

    items = sorted(items, reverse=True)

    space = []

    for item in items:

        best = -1
        best_space = float("inf")

        for i in range(len(space)):
            if space[i] >= item and space[i] - item < best_space:
                best = i
                best_space = space[i] - item

        if best == -1:
            space.append(capacity - item)
        else:
            space[best] -= item

    return len(space)


# ---------- Simulation ----------

TRIALS = 100
ITEMS = 50

total_nfd = 0
total_ffd = 0
total_bfd = 0

random.seed(100)

for _ in range(TRIALS):

    data = [random.uniform(0.1, 0.9) for _ in range(ITEMS)]

    total_nfd += next_fit_decreasing(data)
    total_ffd += first_fit_decreasing(data)
    total_bfd += best_fit_decreasing(data)

print("Average Results after", TRIALS, "Trials\n")

print("Next Fit Decreasing (NFD) :", round(total_nfd / TRIALS, 2))
print("First Fit Decreasing (FFD):", round(total_ffd / TRIALS, 2))
print("Best Fit Decreasing (BFD) :", round(total_bfd / TRIALS, 2))
```
