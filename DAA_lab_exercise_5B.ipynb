```python
import random

comparison_count = 0

def min_max_second(arr, low, high):
    global comparison_count

    # Base case: one element
    if low == high:
        return arr[low], None, arr[low], None

    # Base case: two elements
    if high == low + 1:
        comparison_count += 1
        if arr[low] < arr[high]:
            return arr[low], arr[high], arr[high], arr[low]
        else:
            return arr[high], arr[low], arr[low], arr[high]

    # Divide
    mid = (low + high) // 2
    lmin, lsmin, lmax, lsmax = min_max_second(arr, low, mid)
    rmin, rsmin, rmax, rsmax = min_max_second(arr, mid + 1, high)

    # Find overall minimum
    comparison_count += 1
    if lmin < rmin:
        overall_min = lmin
        candidates = [x for x in [lsmin, rmin] if x is not None]
    else:
        overall_min = rmin
        candidates = [x for x in [rsmin, lmin] if x is not None]

    if len(candidates) == 1:
        second_min = candidates[0]
    else:
        comparison_count += 1
        second_min = min(candidates)

    # Find overall maximum
    comparison_count += 1
    if lmax > rmax:
        overall_max = lmax
        candidates = [x for x in [lsmax, rmax] if x is not None]
    else:
        overall_max = rmax
        candidates = [x for x in [rsmax, lmax] if x is not None]

    if len(candidates) == 1:
        second_max = candidates[0]
    else:
        comparison_count += 1
        second_max = max(candidates)

    return overall_min, second_min, overall_max, second_max


# Driver Code
arr = [random.randint(1, 100) for _ in range(20)]

comparison_count = 0
mn, smn, mx, smx = min_max_second(arr, 0, len(arr) - 1)

print("Array:", arr)
print("Minimum:", mn)
print("Second Minimum:", smn)
print("Maximum:", mx)
print("Second Maximum:", smx)
print("Total Comparisons:", comparison_count)
```
